##Ejercicio 1. Completar Newton (Algoritmo 2.3), unos 20 minutos



```python
def newton_raphson(f, df, p0, TOL=1e-6, N0=50):
    """Algoritmo 2.3. Devuelve (p, historia), con historia = [p0, p1, ..., p]."""
    historia = [p0]
    i = 1
    while i <= N0:
        d = df(p0)
        if d == 0:
            raise ZeroDivisionError(f"f'(p0) = 0 en p0 = {p0}: la tangente es horizontal")
        p = p0 - f(p0) / df(p0)                          # COMPLETE: Paso 3, usando d
        historia.append(p)             # COMPLETE
        if abs(p - p0) < TOL:                          # COMPLETE: Paso 4
            return p, historia
        i = i + 1
        p0 = p                         # COMPLETE: Paso 6
    raise RuntimeError(f"El método fracasó después de {N0} iteraciones")
```

```python
# Definimos la función:
# f(x) = x^3 + 4x^2 - 10

f = lambda x: x**3 + 4*x**2 - 10

# Definimos su derivada:
# f'(x) = 3x^2 + 8x

df = lambda x: 3*x**2 + 8*x

# Aplicamos Newton-Raphson
p, historia = newton_raphson(f, df, 1.5)

print("Raíz aproximada:", p)
print("Historia:", historia)
```

##Ejercicio 2. Escribir la secante (Algoritmo 2.4), unos 45 minutos

```python
def secante(f, p0, p1, TOL=1e-6, N0=50):
    """Algoritmo 2.4. Devuelve (p, historia), con historia = [p0, p1, p2, ..., p]."""
    # Paso 1: i = 2; q0 = f(p0); q1 = f(p1); historia empieza con p0 y p1.
    i = 2
    q0 = f(p0)
    q1 = f(p1)
    historia = [p0, p1]

    # Paso 2: mientras i <= N0, haga los pasos 3 a 6.
    while i <= N0:
        # Si q1 == q0, la secante es horizontal: lance un ZeroDivisionError con un mensaje claro.
        if q1 == q0:
            raise ZeroDivisionError("División por cero: q1 == q0, la recta secante es horizontal.")

        # Paso 3: calcule p y agréguelo a la historia.
        p = p1 - q1 * (p1 - p0) / (q1 - q0)
        historia.append(p)

        # Paso 4: si |p - p1| < TOL, devuelva (p, historia).
        if abs(p - p1) < TOL:
            return p, historia

        # Paso 5: i = i + 1.
        i += 1

        # Paso 6: corra los puntos (p0, q0) <- (p1, q1) y (p1, q1) <- (p, f(p)).
        #         Es la única evaluación nueva de f en la iteración.
        p0, q0 = p1, q1
        p1, q1 = p, f(p)

    raise RuntimeError(f"El método fracasó después de {N0} iteraciones.")
```

```python
# Conteo de las Evaluaciones de f
llamadas = []            # cada evaluación de f agrega aquí su x

def f(x):
    llamadas.append(x)
    return x**3 + 4*x**2 - 10

llamadas.clear()         # vacíe la lista antes de cada prueba
p, historia = secante(f, 1.0, 2.0, TOL=1e-6)
print(p, len(historia) - 2, len(llamadas))

print("Raíz aproximada:", p)
print("Historia:", historia)
```

##Ejercicio 3. ¿Qué tasa de interés cobra el banco? Unos 30 minutos

```python
def saldo_final(i, monto=20000.0, cuota=450.0, meses=60):
    """Saldo después de pagar todas las cuotas, con interés mensual i (0.01 es 1 %)."""
    saldo = monto
    for mes in range(meses):
        saldo = saldo * (1 + i) - cuota            # COMPLETE: interés del mes y pago de la cuota
    return saldo
```

```python
#1. Complete la función y pruébela con un caso que pueda calcular a mano: sin interés, saldo_final(0) debe ser 20000−60⋅450.
print(saldo_final(0))
```

```python
# 2. Encontrar i con el método de la secante
i, historia = secante(saldo_final, 0.005, 0.02, TOL=1e-10)

print("Resultado de i:", i)

# Convertir a porcentaje
porcentaje_i_mensual = i * 100

print(f"Tasa de interés mensual: {porcentaje_i_mensual:.4f}%")
```

```python
#3. Calcule la tasa efectiva anual.
tasa_efectiva_anual = (1 + i)**12 - 1
print("Tasa efectiva anual: " , tasa_efectiva_anual)
print(f"Tasa efectiva anual porcentuak: {(tasa_efectiva_anual * 100):.2f}%")
```
